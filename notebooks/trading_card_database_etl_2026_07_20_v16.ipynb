```python
import pandas as pd
import mysql.connector

db = mysql.connector.connect(
    host="localhost",
    user="root",
    password="password",
    database="trading_card_database"
)
cursor = db.cursor()

target_cols = ['subtype_a', 'subtype_b', 'subtype_c', 'subtype_d', 'subtype_e']

for index, row in df.iterrows():
    for col in target_cols:
        val = row[col]

        if pd.isna(val):
            continue

        query_parent = "SELECT subtype_id FROM subtype WHERE subtype = %s"
        cursor.execute(query_parent, (val,))
        result = cursor.fetchone()

        if result:
            parent_id = result[0]
            insert_child = "INSERT INTO child_table (parent_id, df_index) VALUES (%s, %s)"
            cursor.execute(insert_child, (parent_id, index))

db.commit()
cursor.close()
db.close()
```
